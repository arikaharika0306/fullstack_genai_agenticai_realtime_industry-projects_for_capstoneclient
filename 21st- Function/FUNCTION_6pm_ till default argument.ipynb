# FUNCTION ARGUMENT IN DETAIL 

# POSITIONAL ARGUMENT

```python
def person(name,age): 
    print(name)
    print(age)
    
person('nit')
```

```python
def person(name,age): 
    print(name)
    print(age)
    
person(23)
```

```python
def person(age, name): 
    print(name)
    print(age)
    
person(23)
```

```python
def person(name,age): 
    print(name)
    print(age)
    
person('nit', 23)
```

```python
def person(name,age): 
    print(name)
    print(age)
    
person('nit', 23)
```

```python
def person(name,age): 
    print(name)
    print(age)
    
person(23, 'nit')
```

```python
def person(name,age): 
    print(name)
    print(age + 1)
    
person('nit', 23)
```

```python
def person(name,age): 
    print(name)
    print(age + 1)
    
person(23, 'nit')
```

# KEYWORD ARGUMENT 

```python
def person(name,age): 
    print(name)
    print(age + 1)
    
person(age = 23, name='nit')
```

```python
def person(name,age, mob): 
    print(name)
    print(age + 1)
    print(mob)
    
person(age = 23, name='nit', 4567)
```

```python
def person(name,age, mob): 
    print(name)
    print(age + 1)
    print(mob)
    
person(age = 23, name='nit', mob= 4567)
```

```python
def person(name,age, mob): 
    print(name)
    print(age )
    print(mob)
    
person('nit', 23,  4567)
```

# default argument 

```python
def person(name,age): 
    
    print(name)
    print(age)
    
person('alex' )
```

```python
def person(name, age = 18): 
    
    print(name)
    print(age)
    
person('alex' )
```
